This notebook explores tipping behavior using the `tips` dataset. We will use statistical methods to determine if factors like smoking status, gender, or the day of the week significantly impact bills and tips.

## 2. Feature Engineering: Calculating Tip Percentage

To better understand tipping behavior, we'll create a new feature: `tip_pct`, which represents the tip amount as a percentage of the total bill.

```python
df.info()
```

After adding the new `tip_pct` column, we can see it reflected in the DataFrame's information. It's a `float64` type, as expected.

```python
df.head(25)
```

Looking at the first 25 rows again, we can now see the calculated `tip_pct` for each entry.

```python
smoker_pct = df[df['smoker'] == 'Yes']['tip_pct']
non_smoker_pct = df[df['smoker'] == 'No']['tip_pct']

t_stat, p_val = stats.ttest_ind(smoker_pct, non_smoker_pct)
print(f'T-statistic: {t_stat:.4f}')
print(f'P-value: {p_val:.4f}')
```

## 2. Hypothesis Testing: Smokers vs. Non-Smokers

**Goal:** Determine if smokers tip a different percentage than non-smokers.

*   **H0 (Null Hypothesis):** No difference in mean tip % between groups.
*   **H1 (Alternative Hypothesis):** There is a significant difference.

```python
# T-test

t_stat, p_val = stats.ttest_ind(smoker_pct, non_smoker_pct)
print(f't-stat : {t_stat}, p-val : {p_val}')
```

Now, we perform the independent samples t-test. The `stats.ttest_ind` function from `scipy` calculates the t-statistic and the p-value.

**Interpretation:**
*   **t-statistic:** Measures the difference between the group means in units of standard error.
*   **p-value:** The probability of observing a t-statistic as extreme as, or more extreme than, the one calculated, assuming the null hypothesis is true.

With a p-value of `0.63189`, which is much greater than a common significance level (e.g., 0.05), we **fail to reject the null hypothesis**. This suggests that there is no statistically significant difference in tip percentages between smokers and non-smokers in this dataset.

```python
male_pct = df[df['sex'] == 'Male']['tip_pct']
female_pct = df[df['sex'] == 'Female']['tip_pct']

t_stat, p_val = stats.ttest_ind(male_pct, female_pct, equal_var=False)
print(f'P-value: {p_val:.4f}')
```

## 3. Hypothesis Testing: Gender Influence

**Goal:** Does the gender of the payer affect the tip percentage?

*   **H0:** Gender has no effect on tip %.

```python
days = ['Thur', 'Fri', 'Sat', 'Sun']
groups = [df[df['day'] == d]['total_bill'] for d in days]

f_stat, p_val = stats.f_oneway(*groups)
print(f'F-statistic: {f_stat:.4f}')
print(f'P-value: {p_val:.4f}')
```

## 4. ANOVA: Comparing Bills Across Days

**Goal:** Test if the average bill amount changes depending on the day of the week.

*   **H0:** The mean total bill is the same for all days.
*   **H1:** At least one day has a significantly different mean bill.

```python
for g in groups:
  print(g.mean(), g.std(), len(g))
```

Before running ANOVA, it's good practice to inspect the descriptive statistics (mean, standard deviation, and count) for each group.

```python
f_stat, p_val = stats.f_oneway(*groups)
print(f'f-stat : {f_stat}, p-val : {p_val}')
```

Now, we perform the one-way ANOVA test using `stats.f_oneway`.

**Interpretation:**
*   **F-statistic:** The ratio of the variance between the groups to the variance within the groups.
*   **p-value:** The probability of observing an F-statistic as extreme as, or more extreme than, the one calculated, assuming the null hypothesis is true.

With a p-value of `0.0424`, which is less than 0.05, we **reject the null hypothesis**. This suggests that there is a statistically significant difference in the mean `total_bill` among at least some of the days of the week.

## 5. Chi-Squared Test: categorical Relationships

We use a Chi-Squared test to see if categorical variables like `Time of Day` and `Smoking Status` are related.

```python
contingency = pd.crosstab(df['time'], df['smoker'])
chi2, p_val, dof, expected = stats.chi2_contingency(contingency)

print(f'Chi2 P-value: {p_val:.4f}')
display(pd.DataFrame(expected, columns=contingency.columns, index=contingency.index))
```

**Hypotheses:**
*   **Null Hypothesis (H0):** Smoking status and meal time are independent (no relationship).
*   **Alternative Hypothesis (H1):** Smoking status and meal time are dependent (there is a relationship).

Now, we perform the Chi-squared test for independence using `stats.chi2_contingency`.

```python
pd.DataFrame(expected, columns=contingency.columns, index=contingency.index)
```

With a p-value of `0.4771`, which is greater than 0.05, we **fail to reject the null hypothesis**. This indicates that there is no statistically significant relationship between smoking status and meal time in this dataset.
