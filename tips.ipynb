```python
import pandas as pd
from scipy import stats
```

```python
df = pd.read_csv('/content/tips.csv')
```

```python
df.head()
```

```python
df.info()
```

```python
df['tip_pct'] = (df['tip'] / df['total_bill']) * 100
```

```python
df.info()
```

```python
df.head(25)
```

```python
# H0 = There is no difference %
# H1 = There is differnce in tip %

smoker_pct = df[df['smoker'] == 'Yes']['tip_pct']
non_smoker_pct = df[df['smoker'] == 'No']['tip_pct']

print(f'Smoker % : {smoker_pct.mean()}, {smoker_pct.std()}')
print(f'Non Smoker % : {non_smoker_pct.mean()}, {non_smoker_pct.std()}')
```

```python
# T-test

t_stat, p_val = stats.ttest_ind(smoker_pct, non_smoker_pct)
print(f't-stat : {t_stat}, p-val : {p_val}')
```

```python
# H0: There is no difference in tip by Gender
# H1: There is differ...
male_pct = df[df['sex'] == 'Male']['tip_pct']
female_pct = df[df['sex'] == 'Female']['tip_pct']

t_stat, p_val = stats.ttest_ind(male_pct, female_pct, equal_var=False)
print(f't-stat : {t_stat}, p-val : {p_val}')
```

```python
print(len(df[df['sex'] == 'Male']))
print(len(df[df['sex'] == 'Female']))
```

```python
groups = []
for d in ['Thur', 'Fri', 'Sat', 'Sun'] :
  groups.append(df[df['day'] == d]['total_bill'])
```

```python
for g in groups:
  print(g.mean(), g.std(), len(g))
```

```python
f_stat, p_val = stats.f_oneway(*groups)
print(f'f-stat : {f_stat}, p-val : {p_val}')
```

```python
contingency = pd.crosstab(df['time'], df['smoker'])
contingency
```

```python
# H0 = Smoking status and meal time has no relationship
# H1 = Both are related.

chi2, p_val, dof, expected = stats.chi2_contingency(contingency)
print(f'chi2 : {chi2}, p-val : {p_val}, dof : {dof}')
```

```python
pd.DataFrame(expected, columns=contingency.columns, index=contingency.index)
```
